## Ejercicio 2

```python
# Genero un arreglo de tiempo de 10 años con 500 mediciones.
t = np.linspace(0, 10, 500)

# Defino el flujo inicial normalizado de la estrella.
F = np.ones(500)

# Agrego ruido gaussiano con desviación estándar de 0.002.
ruido = rng.normal(0, 0.002, 500)
F = F + ruido

# Creo una máscara para el intervalo de tiempo en que ocurre el tránsito.
msk_transito = (t >= 4.5) & (t <= 5.0)

# Reduzco el flujo en 0.02 durante el tránsito.
F[msk_transito] = F[msk_transito] - 0.02

# Calculo el flujo promedio.
promedio = np.mean(F)

promedio
```

```python
# Calculo la desviación estándar del flujo.
desviacion = np.std(F)

desviacion
```

```python
# Determino el flujo mínimo y su índice.
flujo_minimo = np.min(F)
indice_minimo = np.argmin(F)

flujo_minimo, indice_minimo
```

```python
# Determino el tiempo correspondiente al flujo mínimo.
tiempo_minimo = t[indice_minimo]

tiempo_minimo
```

```python
# Selecciono los puntos que cumplen con el criterio de 3 sigma.
msk = F < promedio - 3 * desviacion

msk
```

```python
# Cuento la cantidad de posibles puntos de tránsito.
cantidad_transito = np.sum(msk)

cantidad_transito
```

```python
# Calculo el flujo promedio durante el tránsito y estimo su profundidad.
F_transito = np.mean(F[msk])
delta_F = 1 - F_transito

delta_F
```

```python
# Estimo la razón entre el radio del planeta y el radio de la estrella.
Rp_Rs = np.sqrt(delta_F)

Rp_Rs
```

```python
# Obtengo los tiempos identificados con el criterio de 3 sigma.
tiempos_detectados = t[msk]

tiempos_detectados
```
